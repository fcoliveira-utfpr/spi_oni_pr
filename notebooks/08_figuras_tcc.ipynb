```python
import shutil
import pandas as pd

FIGURAS = {                       # arquivo gravado pelos notebooks -> nome final, na ordem do TCC
    "fig01_mapa_estacoes.png": "Figura01_mapa_estacoes.png",
    "fig02a_regime_mensal_mesorregioes.png": "Figura02a_regime_mensal_mesorregioes.png",
    "fig02b_regime_mensal_destaque.png": "Figura02b_regime_mensal_destaque.png",
    "fig03_homogeneidade_tendencia.png": "Figura03_homogeneidade_tendencia.png",
    "fig04_ajuste_fdp_santa_helena.png": "Figura04_ajuste_fdp.png",
    "fig05a_spi3_destaque.png": "Figura05a_spi3_destaque.png",
    "fig05b_spi12_destaque.png": "Figura05b_spi12_destaque.png",
    "fig06_mapas_ano_seco_umido.png": "Figura06_mapas_ano_seco_umido.png",
    "fig07_oni_fases.png": "Figura07_oni_fases.png",
    "fig08_spi12_fases_enos.png": "Figura08_spi12_fases_enos.png",
    "fig09_dispersao_oni_spi3.png": "Figura09_dispersao_oni_spi3.png",
    "fig10_correlacao_mes_defasagem.png": "Figura10_correlacao_mes_defasagem.png",
    "fig11_mapas_spi_enos.png": "Figura11_mapas_spi_enos.png",
}
TABELAS = {
    "tabela02_chuva_anual_mesorregiao.csv": "Tabela2",
    "tabela03_homogeneidade_tendencia.csv": "Tabela3",
    "tabela04_ajuste_fdp.csv": "Tabela4",
    "tabela05_eventos_spi12_destaque.csv": "Tabela5",
    "tabela06_eventos_enos.csv": "Tabela6",
    "tabela07_correlacao_sazonal.csv": "Tabela7",
    "tabela08_probabilidade_condicional.csv": "Tabela8",
    "tabela09_resumo_mesorregiao.csv": "Tabela9",
}
```

## 1. Figuras

Copia cada figura para `figuras_tcc` com o nome final. Se faltar alguma, o notebook avisa qual etapa rodar.

```python
pasta = SAIDA / "figuras_tcc"
pasta.mkdir(exist_ok=True)
for origem, destino in FIGURAS.items():
    if (SAIDA / origem).exists():
        shutil.copy(SAIDA / origem, pasta / destino)
        print("ok      ", destino)
    else:
        print("FALTANDO", origem, "-> rode o notebook que cria esta figura")
```

## 2. Tabelas

Junta as tabelas numa planilha do Excel, uma aba por tabela. Os números ficam com o ponto decimal; ao colar no
Google Docs, troque por vírgula e arredonde para o número de casas que fizer sentido.

```python
with pd.ExcelWriter(SAIDA / "tabelas_tcc.xlsx") as planilha:
    for arquivo, aba in TABELAS.items():
        if (SAIDA / arquivo).exists():
            pd.read_csv(SAIDA / arquivo).to_excel(planilha, sheet_name=aba, index=False)
            print("ok      ", aba)
        else:
            print("FALTANDO", arquivo)
print("Planilha gravada em", SAIDA / "tabelas_tcc.xlsx")
```

## Lista de conferência antes de colar no TCC2

- [ ] Cada figura tem título **acima** ("Figura N - ...") e fonte **abaixo** ("Fonte: Autoria própria (2027)."),
      no padrão da UTFPR;
- [ ] Cada figura e tabela é citada no texto **antes** de aparecer;
- [ ] Unidades nos eixos e nas colunas (mm, °C, %);
- [ ] Vírgula decimal nas tabelas;
- [ ] Os números citados no texto batem com os das tabelas;
- [ ] Sumário, lista de ilustrações e lista de tabelas atualizados no Google Docs.
