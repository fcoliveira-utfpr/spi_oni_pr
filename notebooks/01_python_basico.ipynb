## 1. Variáveis e contas

Uma **variável** é um nome que guarda um valor. O sinal `=` **guarda** o valor da direita no nome da esquerda (não
é o "igual" da matemática). `print` mostra valores na tela.

Uma curiosidade útil: **1 mm de chuva = 1 litro de água por metro quadrado**.

```python
chuva = 25.0                 # chuva de um dia, em mm (o ponto separa os decimais, como na calculadora)
area_ha = 1.0                # área de 1 hectare

litros_por_m2 = chuva * 1    # 1 mm = 1 L/m²
metros2 = area_ha * 10000    # 1 ha = 10.000 m²
litros_total = litros_por_m2 * metros2

print("Chuva do dia:", chuva, "mm")
print("Água em 1 ha:", litros_total, "litros")
print("Água em 1 ha:", litros_total / 1000, "m³")   # 1 m³ = 1.000 litros
```

## 2. Listas e o laço `for`

Uma **lista** guarda vários valores em ordem, entre colchetes `[ ]`. O laço `for` repete um bloco de código para
cada item da lista. O bloco repetido é o que está **recuado** (4 espaços) abaixo do `for`.

```python
chuvas = [0.0, 12.4, 3.1, 0.0, 25.8, 7.2, 0.0]   # chuva (mm) de uma semana, um valor por dia

print("Chuva do primeiro dia:", chuvas[0])   # em Python, a contagem começa em 0
print("Chuva do último dia:", chuvas[-1])    # -1 = o último item
print("Número de dias:", len(chuvas))        # len = tamanho da lista

total = 0                                    # começa a soma em zero
for valor in chuvas:                         # "para cada valor na lista chuvas"
    total = total + valor                    # vai somando, um dia de cada vez
print("Chuva da semana:", total, "mm")

print("Conferindo com a função sum():", sum(chuvas), "mm")
```

## 3. Decisões com `if`

`if` executa um bloco **só se** a condição for verdadeira; `elif` ("senão, se") e `else` ("senão") tratam os
outros casos. Em meteorologia, um dia é considerado **chuvoso** quando chove **1 mm ou mais**.

```python
chuvas = [0.0, 12.4, 0.6, 0.0, 25.8, 7.2, 0.2]

for valor in chuvas:
    if valor >= 1.0:                 # >= quer dizer "maior ou igual"
        tipo = "chuvoso"
    else:
        tipo = "seco"
    print(valor, "mm ->", tipo)
```

## 4. Funções

Uma **função** é uma receita com nome: recebe valores, faz as contas e devolve um resultado com `return`. Ela evita
repetir o mesmo código.

A função abaixo classifica um valor de SPI pelas faixas de McKee, Doesken e Kleist (1993), as mesmas da Tabela 1 do
TCC. O `if` / `elif` testa as faixas de cima para baixo e **para na primeira que for verdadeira**.

```python
def classe_spi(valor):
    """Recebe um valor de SPI e devolve a classe de McKee et al. (1993)."""
    if valor >= 2.0:
        return "Extremamente úmido"
    elif valor >= 1.5:
        return "Muito úmido"
    elif valor >= 1.0:
        return "Moderadamente úmido"
    elif valor > -1.0:
        return "Normal"
    elif valor > -1.5:
        return "Moderadamente seco"
    elif valor > -2.0:
        return "Severamente seco"
    else:
        return "Extremamente seco"


# testando a função com alguns valores
for spi in [2.3, 1.2, 0.4, -0.99, -1.0, -1.7, -2.5]:
    print(spi, "->", classe_spi(spi))
```

## 5. Dicionários

Um **dicionário** guarda pares **chave: valor**, entre chaves `{ }`. Funciona como uma agenda: você procura pela
chave (o nome) e encontra o valor (o telefone). Aqui, o código de cada estação em destaque.

```python
estacoes = {
    "Santa Helena": "2454012",
    "Cascavel": "2453056",
    "Foz do Iguaçu": "2554002",
    "Mercedes": "2454015",
}

print("Código de Cascavel:", estacoes["Cascavel"])     # consulta pela chave

for nome, codigo in estacoes.items():                  # .items() percorre os pares
    print(nome, "tem o código", codigo)
```

## 6. Tabelas com pandas

A biblioteca **pandas** trabalha com tabelas (como uma planilha). `import` carrega a biblioteca; `pd` é o apelido
que todo mundo usa. `pd.read_csv` lê um arquivo de texto separado por vírgulas.

O arquivo `chuva_diaria_destaque.csv` tem a chuva de **cada dia** das quatro estações em destaque, de 1981 a 2020.
As colunas são: código da estação (`cod`), data, chuva (mm) e `preenchido` (0 = observado no pluviômetro;
1 = falha preenchida com a grade de Xavier et al., 2022).

```python
import pandas as pd

# dtype={"cod": str} diz que o código da estação é um texto (e não um número)
diaria = pd.read_csv(DADOS + "chuva_diaria_destaque.csv", dtype={"cod": str})

print("Linhas e colunas:", diaria.shape)
diaria.head()        # mostra as 5 primeiras linhas
```

Para trabalhar com datas, convertemos a coluna `data` (que veio como texto) para o tipo data. Depois separamos o
ano e o mês em colunas próprias. Em seguida, **filtramos** só Santa Helena: a expressão entre colchetes escolhe as
linhas em que a condição é verdadeira.

```python
diaria["data"] = pd.to_datetime(diaria["data"])     # texto -> data
diaria["ano"] = diaria["data"].dt.year              # .dt.year pega o ano da data
diaria["mes"] = diaria["data"].dt.month             # .dt.month pega o mês

sh = diaria[diaria["cod"] == "2454012"]             # só as linhas de Santa Helena
print("Dias de Santa Helena:", len(sh))
print("Maior chuva em um dia:", sh["chuva_mm"].max(), "mm")

dia_max = sh[sh["chuva_mm"] == sh["chuva_mm"].max()]   # a linha do dia de maior chuva
dia_max
```

Agora a **chuva de cada mês de 2019** em Santa Helena. Primeiro do jeito mais simples, com um laço `for` que passa
pelos 12 meses e soma a chuva dos dias de cada um.

```python
sh2019 = sh[sh["ano"] == 2019]                      # só o ano de 2019

meses = []                                          # listas vazias para guardar os resultados
totais = []
for m in range(1, 13):                              # range(1, 13) = 1, 2, ..., 12
    dias_do_mes = sh2019[sh2019["mes"] == m]
    total = dias_do_mes["chuva_mm"].sum()           # soma a chuva dos dias do mês
    meses.append(m)                                 # .append acrescenta um item no fim da lista
    totais.append(total)
    print("Mês", m, ":", round(total, 1), "mm")

print("Total do ano:", round(sum(totais), 1), "mm")
```

O pandas faz a mesma conta numa linha só, com `groupby`: ele **junta** as linhas que têm o mesmo mês e **soma** a
chuva dentro de cada grupo. Confira que os números batem com os do laço acima.

```python
mensal2019 = sh2019.groupby("mes")["chuva_mm"].sum()
mensal2019.round(1)
```
